```python
def recommend_food(food_name):

    food_index = recommended_foods[
        recommended_foods["food"] == food_name
    ].index[0]

    filtered_position = list(
        recommended_foods.index
    ).index(food_index)

    distances, indices = model.kneighbors(
        [X_scaled[filtered_position]]
    )

    recommendations = recommended_foods.iloc[
        indices[0]
    ][
        ["food", "Protein", "Sugars", "Fat"]
    ]

    recommendations = recommendations.drop_duplicates(
        subset="food"
    )

    recommendations = recommendations[
        recommendations["food"] != food_name
    ]

    return recommendations
```

```python
def recommend_food(food_name):

    try:

        # FIND FOOD INDEX
        food_index = recommended_foods[
            recommended_foods["food"] == food_name
        ].index[0]

        # GET POSITION
        filtered_position = list(
            recommended_foods.index
        ).index(food_index)

        # FIND SIMILAR FOODS
        distances, indices = model.kneighbors(
            [X_scaled[filtered_position]]
        )

        # GET RECOMMENDATIONS
        recommendations = recommended_foods.iloc[
            indices[0]
        ][
            [
                "food",
                "Caloric Value",
                "Protein",
                "Sugars",
                "Fat",
                "Nutrition Density"
            ]
        ]

        # REMOVE DUPLICATES
        recommendations = recommendations.drop_duplicates(
            subset="food"
        )

        # REMOVE INPUT FOOD
        recommendations = recommendations[
            recommendations["food"] != food_name
        ]

        return recommendations

    except:
        print("Food not found!")
```

```python
def explain_food(food):

    selected = recommended_foods[
        recommended_foods["food"] == food
    ]

    if len(selected) == 0:
        print("Food not found!")
        return

    print("\nRecommended because:\n")

    if selected["Sugars"].values[0] < 10:
        print("- Low sugar")

    if selected["Fat"].values[0] < 10:
        print("- Low fat")

    if selected["Protein"].values[0] > 15:
        print("- High protein")

    if selected["Nutrition Density"].values[0] > 0.5:
        print("- High nutrition density")
```

```python
def get_daily_calories(activity_level):

    if activity_level == "Sedentary":
        return 1800

    elif activity_level == "Moderately Active":
        return 2200

    elif activity_level == "Highly Active":
        return 2800

    else:
        return 2000
```
